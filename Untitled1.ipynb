```python
import pandas as pd
import numpy as np
import os

# --- KONFIGURACJA ŚCIEŻEK ---
BASE_DIR = r"C:\Users\Bartek\Documents\aaaaaaUCZELNIA\Wpr.doEksp.danych\projekt_choroby"

# Definicje podfolderów
RAW_DIR = os.path.join(BASE_DIR, "data_raw")
CLEAN_DIR = os.path.join(BASE_DIR, "data_clean")

# Upewnijmy się, że folder wyjściowy istnieje
os.makedirs(CLEAN_DIR, exist_ok=True)

# Mapa plików
FILES = {
    'MEASLES': os.path.join(RAW_DIR, 'measles_data.csv'),
    'PERTUSSIS': os.path.join(RAW_DIR, 'pertussis_data.csv'),
    'SALMONELLOSIS': os.path.join(RAW_DIR, 'salmonellosis_data.csv'),
    'TUBERCULOSIS': os.path.join(RAW_DIR, 'tuberculosis_data.csv'),
    
    # ZMIANA: Szukamy kontekstu również w folderze data_raw (tam gdzie inne CSV)
    'CONTEXT': os.path.join(RAW_DIR, 'dataset_final_bartek.csv')
}

# --- DIAGNOSTYKA ---
print(f"📂 Katalog roboczy: {BASE_DIR}")
print("🔍 Weryfikacja plików:")

missing = []
for name, path in FILES.items():
    if os.path.exists(path):
        print(f"   ✅ {name:<15}: OK")
    else:
        print(f"   ❌ {name:<15}: BRAK -> {path}")
        missing.append(name)

if not missing:
    print("\n🚀 Wszystkie pliki znalezione w 'data_raw'. Możesz przetwarzać!")
else:
    print("\n⚠️  NADAL BRAKUJE PLIKÓW.")
    print("   Upewnij się, że plik 'dataset_final_bartek.csv' został skopiowany do folderu:")
    print(f"   📂 {RAW_DIR}")
```

```python
def load_context_data(path):
    """Wczytuje dane o strukturze wieku i szczepieniach ze starego zbioru."""
    try:
        # Uwaga: Plik źródłowy Bartka miał średniki!
        df = pd.read_csv(path, sep=';')
        # Wyciągamy unikalne wiersze kontekstowe
        return df[['Year', 'Country', 'Pop_Structure_0_14_Pct', 'Vaccination_Coverage_Pct']].drop_duplicates()
    except FileNotFoundError:
        print("⚠️ Brak pliku kontekstowego. Wyniki będą bez danych o szczepieniach.")
        return pd.DataFrame(columns=['Year', 'Country', 'Pop_Structure_0_14_Pct', 'Vaccination_Coverage_Pct'])

def process_disease_data(file_path, disease_name, context_df, is_monthly=False):
    """
    Wykonuje pełny ETL:
    1. Wczytanie (Load)
    2. Agregacja miesięczna -> roczna (Transform)
    3. Imputacja populacji (Transform)
    4. Integracja z kontekstem (Merge)
    5. Rekalkulacja zapadalności (Incidence)
    """
    # 1. Wczytanie surowych danych (format kacpra: przecinki)
    df = pd.read_csv(file_path)
    
    if 'RegionName' in df.columns:
        df.rename(columns={'RegionName': 'Country'}, inplace=True)

    # 2. Agregacja (dla Odry)
    if is_monthly and 'Month' in df.columns:
        # Suma przypadków, średnia populacja w roku
        df = df.groupby(['Year', 'Country']).agg({
            'Cases': 'sum',
            'Population': 'mean'
        }).reset_index()
    else:
        df = df[['Year', 'Country', 'Cases', 'Population']]

    # 3. Imputacja Populacji (Interpolacja liniowa)
    df['Population'] = df['Population'].replace(0, np.nan)
    # Wypełnianie dziur między latami dla każdego kraju osobno
    df['Population'] = df.groupby('Country')['Population'].transform(
        lambda x: x.interpolate(limit_direction='both')
    )

    # 4. Merge z kontekstem (Left Join)
    df_merged = pd.merge(df, context_df, on=['Year', 'Country'], how='left')

    # 5. Przeliczenie Incidence (Standard: na 100k mieszkańców)
    df_merged['Incidence'] = np.where(
        df_merged['Population'] > 0,
        (df_merged['Cases'] / df_merged['Population']) * 100000,
        np.nan
    )

    # Formatowanie końcowe
    df_merged['Disease'] = disease_name
    df_merged['Incidence'] = df_merged['Incidence'].round(2)
    df_merged['Population'] = df_merged['Population'].round(0)
    
    # Uporządkowanie kolumn
    final_cols = [
        'Year', 'Country', 'Disease', 'Cases', 'Population', 'Incidence',
        'Pop_Structure_0_14_Pct', 'Vaccination_Coverage_Pct'
    ]
    
    # Zabezpieczenie na wypadek braku kolumn po merge
    for col in final_cols:
        if col not in df_merged.columns:
            df_merged[col] = np.nan
            
    return df_merged[final_cols].sort_values(['Year', 'Country'])

print("✅ Funkcje ETL zdefiniowane.")
```

```python
# 1. Załaduj kontekst
context_df = load_context_data(FILES['CONTEXT'])

# 2. Zdefiniuj zadania
tasks = [
    ('Measles', FILES['MEASLES'], True),       # True = dane miesięczne
    ('Pertussis', FILES['PERTUSSIS'], False),
    ('Salmonellosis', FILES['SALMONELLOSIS'], False),
    ('Tuberculosis', FILES['TUBERCULOSIS'], False)
]

print("🚀 Rozpoczynam przetwarzanie i standaryzację formatu...")

for disease, path, monthly_flag in tasks:
    try:
        # Przetwarzanie
        df_clean = process_disease_data(path, disease, context_df, is_monthly=monthly_flag)
        
        # Nazwa pliku wyjściowego
        output_file = f"{disease.lower()}_clean.csv"
        
        # --- ZAPIS (KLUCZOWE: Standard CSV) ---
        # sep=',' (przecinek), decimal='.' (kropka), utf-8
        df_clean.to_csv(output_file, sep=',', decimal='.', index=False, encoding='utf-8')
        
        print(f"   ✅ {disease}: Zapisano '{output_file}' (Wierszy: {len(df_clean)})")
        
    except Exception as e:
        print(f"   ❌ Błąd przy {disease}: {e}")

print("\n🎉 Zakończono. Pliki są gotowe do analizy.")
```

```python
import pandas as pd
import numpy as np
import os

# Konfiguracja plików wejściowych
FILES = [
    'measles_clean.csv',
    'pertussis_clean.csv',
    'salmonellosis_clean.csv',
    'tuberculosis_clean.csv'
]
OUTPUT_FILE = 'dataset_final_bartek.csv'

data_frames = []

print("🚀 Rozpoczynam integrację datasetu...")

for file_path in FILES:
    if os.path.exists(file_path):
        try:
            # 1. Wczytanie danych (jawny separator przecinka)
            df = pd.read_csv(file_path, sep=',')
            
            # 2. Weryfikacja i rzutowanie typów danych
            df['Year'] = df['Year'].astype(int)
            df['Country'] = df['Country'].astype(str).str.strip()
            df['Disease'] = df['Disease'].astype(str).str.strip()
            
            # Konwersja kolumn numerycznych (NaN pozostają bez zmian)
            numeric_cols = [
                'Cases', 'Population', 'Incidence', 
                'Pop_Structure_0_14_Pct', 'Vaccination_Coverage_Pct'
            ]
            for col in numeric_cols:
                df[col] = pd.to_numeric(df[col], errors='coerce')

            # 3. Usuwanie duplikatów (pełne wiersze)
            initial_count = len(df)
            df.drop_duplicates(inplace=True)
            if len(df) < initial_count:
                print(f"   ⚠️ Usunięto {initial_count - len(df)} duplikatów w pliku {file_path}")

            data_frames.append(df)
            print(f"   ✅ Załadowano {file_path}: {len(df)} wierszy")
            
        except Exception as e:
            print(f"   ❌ Błąd przetwarzania {file_path}: {e}")
    else:
        print(f"   ❌ Nie znaleziono pliku: {file_path}")

# 4. Łączenie i finalizacja
if data_frames:
    final_df = pd.concat(data_frames, ignore_index=True)
    
    # 5. Sortowanie hierarchiczne
    final_df.sort_values(by=['Country', 'Disease', 'Year'], inplace=True)
    
    # 6. Zapis do pliku wynikowego (CSV, UTF-8, przecinek, bez indeksu)
    final_df.to_csv(OUTPUT_FILE, sep=',', encoding='utf-8', index=False)
    
    # Podsumowanie dla użytkownika
    print("\n" + "="*50)
    print(f"🎉 ZAKOŃCZONO SUKCESEM!")
    print(f"💾 Zapisano plik: {OUTPUT_FILE}")
    print(f"📊 Wymiary datasetu: {final_df.shape}")
    print(f"🔍 Podgląd typów danych:\n{final_df.dtypes}")
    print("="*50)
    print(final_df.head().to_markdown(index=False, numalign="left", stralign="left"))
else:
    print("❌ Przerwano: Brak danych do połączenia.")
```
